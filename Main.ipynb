```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("data.csv")

all_types = pd.concat([
    df[["Type 1", "Total",'HP',"Attack","Defense","Sp. Atk","Sp. Def","Speed","Generation","Legendary"]].rename(columns={"Type 1": "Type"}),
    df[["Type 2", "Total",'HP',"Attack","Defense","Sp. Atk","Sp. Def","Speed","Generation","Legendary"]].rename(columns={"Type 2": "Type"})
]).dropna()

type_stats = all_types.groupby("Type")[
    ["HP", "Attack", "Defense", "Sp. Atk", "Sp. Def", "Speed"]
].mean()
#type_count = df["Type 1"].value_counts(ascending=True)

avg_total_by_type = all_types.groupby("Type")["Total"].mean().sort_values()

avg_hp_by_type = all_types.groupby("Type")["HP"].mean().sort_values()
avg_attack_by_type = all_types.groupby("Type")["Attack"].mean().sort_values()
avg_defense_by_type = all_types.groupby("Type")["Defense"].mean().sort_values()
avg_sp_atk_by_type = all_types.groupby("Type")["Sp. Atk"].mean().sort_values()
avg_sp_def_by_type = all_types.groupby("Type")["Sp. Def"].mean().sort_values()
avg_speed_by_type = all_types.groupby("Type")["Speed"].mean().sort_values()

avg_gen_by_type = all_types.groupby("Type")["Generation"].mean().sort_values()
avg_legendary_by_type = all_types.groupby("Type")["Legendary"].mean().sort_values()
```

```python
plt.barh(avg_hp_by_type.index, avg_hp_by_type.values)
plt.show()
```
